```python
import pandas as pd
import os
import glob


def load_and_combine_stocks(base_folder):
    stocks_folder = os.path.join(base_folder, "stocks")

    # Check if folder exists to avoid errors
    if not os.path.exists(stocks_folder):
        print(f"ERROR: The folder {stocks_folder} does not exist.")
        print("Did you move the stock files into a 'stocks' subfolder?")
        return None

    # Search for CSVs inside that specific subfolder
    file_list = glob.glob(os.path.join(stocks_folder, "*.csv"))

    all_data = []
    print(f"Looking in: {stocks_folder}")
    print(f"Found {len(file_list)} stock files.")

    for file in file_list:
        filename = os.path.basename(file)
        ticker = filename.replace(".csv", "").replace(".xlsx", "")

        # Read the file
        try:
            df = pd.read_csv(file)
            df["stock"] = ticker  # Add the label
            all_data.append(df)
            print(f"-> Loaded {ticker}")
        except Exception as e:
            print(f"Failed to load {ticker}: {e}")

    if all_data:
        combined_df = pd.concat(all_data, ignore_index=True)
        # Force column names to lowercase to avoid "Date" vs "date" issues
        combined_df.columns = combined_df.columns.str.lower()
        return combined_df
    else:
        print("No data found!")
        return pd.DataFrame()


path_to_raw = "../data/raw"

stock_df = load_and_combine_stocks(path_to_raw)

if stock_df is not None and not stock_df.empty:
    print("\n--- SUCCESS ---")
    print(stock_df.info())
```

```python
stock_df['date'] = pd.to_datetime(stock_df['date'])
```

```python
# --- 2. Save to Processed Folder ---
# We define the path to the processed folder
processed_path = "../data/processed"
```

```python
stock_df.to_csv(f"{processed_path}/stock_prices_combined.csv", index=False)
```
